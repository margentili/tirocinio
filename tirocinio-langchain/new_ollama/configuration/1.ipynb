```python
import json
from pathlib import Path

dati = json.loads(Path("ferie_archivio.json").read_text(encoding="utf-8"))
for d in dati:
    d["tipo"] = "ferie"
Path("ferie_archivio.json").write_text(json.dumps(dati, ensure_ascii=False, indent=2), encoding="utf-8")
print("normalizzati", len(dati))
```

```python
import json
from pathlib import Path
p = Path("rumore/iban/fatture_archivio.json")
dati = json.loads(p.read_text(encoding="utf-8"))
for d in dati:
    d["tipo"] = "iban"
p.write_text(json.dumps(dati, ensure_ascii=False, indent=2), encoding="utf-8")
print("normalizzati", len(dati))
```

```python
import json, os

OUTPUT_DIR = "rumore/iban"
os.makedirs(OUTPUT_DIR, exist_ok=True)

with open("rumore/iban/fatture_archivio.json", "r", encoding="utf-8") as f:
    emails = json.load(f)

for item in emails:
    file_name = item.get("file")
    percorso_completo = os.path.join(OUTPUT_DIR, file_name)

    contenuto_testo = (
        f"DA: {item.get('da')}\n"
        f"OGGETTO: {item.get('oggetto')}\n"
        f"TIPO: {item.get('tipo')}\n---\n"
        f"{item.get('corpo')}\n"
    )

    with open(percorso_completo, "w", encoding="utf-8") as out:
        out.write(contenuto_testo)

print(f"{len(emails)} file creati in '{OUTPUT_DIR}'.")
```
